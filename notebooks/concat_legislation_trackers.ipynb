# Concatenate ACLU Legislation Trackers (2020–2026)

Loads and concatenates all `legislation-tracker_20XX.csv` files with a `year` column.

```python
import pandas as pd
from pathlib import Path
```

```python
ACLU_DIR = Path('.').resolve().parent if '__file__' not in dir() else Path(__file__).resolve().parent
files = sorted(ACLU_DIR.glob('legislation-tracker_20*.csv'))
print(f"Found {len(files)} files:")
for f in files:
    print(f"  {f.name}")
```

```python
dfs = []
for f in files:
    year = int(f.stem.split('_')[-1])  # e.g. legislation-tracker_2023 -> 2023
    df = pd.read_csv(f)
    df['year'] = year
    # 2022 has no Status Detail; align schema
    if 'Status Detail' not in df.columns:
        df['Status Detail'] = pd.NA
    # Normalize column names (2022 has trailing spaces)
    df.columns = df.columns.str.strip()
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)
print(f"Total rows: {len(combined)}")
print(f"Columns: {list(combined.columns)}")
combined.head()
```

```python
out_path = ACLU_DIR / 'legislation-tracker_2020_2026.csv'
combined.to_csv(out_path, index=False)
print(f"Saved to {out_path}")
```
